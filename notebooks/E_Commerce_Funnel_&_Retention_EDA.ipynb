## Customer Retention



```python
# Count orders per user; a user with more than one order is a repeat customer
orders_per_user = df_orders.groupby('user_id')['order_id'].nunique().reset_index(name='order_count')

repeat_customers = (orders_per_user['order_count'] > 1).sum()
total_customers = df_orders['user_id'].nunique()
new_customers = total_customers - repeat_customers

repeat_pct = (repeat_customers / total_customers) * 100
new_pct = (new_customers / total_customers) * 100

values = [repeat_customers,new_customers]
labels = [f'Repeat Customers\n{repeat_pct:.2f}%',f'New Customers\n{new_pct:.2f}%']

plt.pie(values,labels=labels,colors=['salmon','cornflowerblue'])
plt.title('Customer Retention Breakdown')

plt.show()
```

## Hypothesis Testing

The Customer Value Hypothesis (Independent T-test)

```python
# Ho: There is no statistical difference in spend. Repeat and new customers are equally valuable per transaction
# Ha: Repeat customers spend significantly more per order

df_orders_label = df_orders.merge(orders_per_user,how='left',on='user_id')

one_time_order = df_orders_label.loc[df_orders_label['order_count']==1,'price_usd']
repeat_orders = df_orders_label.loc[df_orders_label['order_count']>1,'price_usd']

t_stats, p_value = ttest_ind(repeat_orders, one_time_order, alternative='greater')


alpha = 0.05
print(f'Pvalue: {p_value}')
if p_value < alpha:
    print("Reject the Null Hypothesis.")
    print("Repeat customers spend significantly more per order. We should invest in retention!")
else:
    print("Fail to reject the Null Hypothesis.")
    print("There is no statistical difference in spend. Repeat and new customers are equally valuable per transaction.")
```

Device type cart abandonment hypothesis (Z-proportion)

```python
# Ho: There is no statistical difference in cart abandonment rate mobile nad destop users have same abandonment rate.
# Ha: Mobile users have more abandonment rate.

df_session_x_view = df_web_page.merge(df_web_sess,how='left',on='website_session_id')

mobile_orders = df_session_x_view[((df_session_x_view['pageview_url']=='/thank-you-for-your-order')&
                                         (df_session_x_view['device_type']=='mobile'))].shape[0]
desktop_orders = df_session_x_view[((df_session_x_view['pageview_url']=='/thank-you-for-your-order')&
                                         (df_session_x_view['device_type']=='desktop'))].shape[0]

mobile_cart_visits = df_session_x_view[((df_session_x_view['pageview_url']=='/cart')&
                                         (df_session_x_view['device_type']=='mobile'))].shape[0]

desktop_cart_visits = df_session_x_view[((df_session_x_view['pageview_url']=='/cart')&
                                         (df_session_x_view['device_type']=='desktop'))].shape[0]

mobile_abandoned = mobile_cart_visits - mobile_orders
desktop_abandoned = desktop_cart_visits - desktop_orders

abandonment_counts = [mobile_abandoned, desktop_abandoned]
total_sessions = [mobile_cart_visits, desktop_cart_visits]

mobile_rate = (mobile_abandoned / mobile_cart_visits) * 100
desktop_rate = (desktop_abandoned / desktop_cart_visits) * 100

z_stat, p_value = proportions_ztest(count=abandonment_counts, nobs=total_sessions, alternative='larger')


print(f"Mobile Abandonment Rate: {mobile_rate:.2f}%")
print(f"Desktop Abandonment Rate: {desktop_rate:.2f}%")
print("-" * 30)
print(f"P-Value: {p_value}")


alpha = 0.05
if p_value < alpha:
    print("\nResult: Reject the Null Hypothesis.")
    print("Mobile users abandon at a significantly higher rate. A mobile UI audit is strongly recommended.")
else:
    print("\nResult: Fail to reject the Null Hypothesis.")
    print("The difference in abandonment is not statistically significant.")
```

## Business Key Insights

1. **The landing page is the single biggest funnel leak bigger than checkout.**
 Around (45%) of our visitors leave the site from the very first page without even looking at our products. This is a largest drop than any single checkout step.

2. **Traffic is good, but sales are low:** 472,871 sessions produced only 32,313 orders that's 6.83% conversion rate. The volume of traffic isn't the problem something between landing and purchase is.

3. **Desktop converts nearly 3x better than mobile.** Desktop converts around  8.50% and The mobile converts around 3.50% may be its checkout process is likely too difficult.

4. **Cart abandonment is high.**: Of 94,953 sessions that reached the cart, 62,640 abandoned it (66% abandonment rate). Past fixes worked initially but then got stable, not improving.

5. **pilot campign**: weak on both 1.08% conversion rate and $3,743 in sales.

7. **"The Original Mr. Fuzzy" has highest sales and the biggest refund risk.** It drives \$1.21M gross sales and around \$677K net profit more than the other three products combined but its 5.11% return rate translates to \$61,838 in refunds the largest refund exposure of any product.
